```python
class animal:
    def make_sound(self):
        print("animals make different sounds")

class cat(animal):
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def info(self):
        print(f"my name is{self.name}and i am {self.age} years old")

    def make_sound(self):
        print("meow meow")

class dog(animal):
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def info(self):
        print(f"my name is{self.name}and i am {self.age} years old")

    def make_sound(self):
        print("woof woof")

cat=cat("kitty", 2)
dog=dog("fluffy", 3)

cat.info()
cat.make_sound()
```

```python
class Dog(animal):
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def make_sound(self):
        print("woof woof")


class Cat(animal):
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def make_sound(self):
        print("meow meow")


my_dog = Dog("fluffy", 3)
my_cat = Cat("kitty", 2)

for entity in (my_dog, my_cat):
    entity.make_sound()
```

```python
class countdowniterater:
    def __init__(self, start):
        self.start = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.start <= 0:
            raise StopIteration
        else:
            current = self.start
            self.start -= 1
            return current
count1=countdowniterater(3)
count2=countdowniterater(5)

for i in count1:
    print(i)
for j in count2:
    print(j)
```

```python
items=[[10,20,30],"MCA",countdowniterater(3)]
for goroup in items:
    for item in goroup:
        print(item)
```
