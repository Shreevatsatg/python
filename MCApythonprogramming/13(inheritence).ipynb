```python
class parent1:
    def feature_one(self):
        print("feature one is working")

    def feature_two(self):
        print("feature two is working")

class child(parent1):
    def feature_three(self):
        print("feature three is working")

c=child()
c.feature_one()
c.feature_two()
c.feature_three()
```

```python



class prey:
    def flee(self):
        print("fleeing from danger")

class predator:
    def hunt(self):
        print("hunting for food")
        
class fish(prey,predator):
    def fee(self):
        print("fish fleeing from danger and hunting for food")
        
fish=fish()
fish.fee()
fish.flee()
fish.hunt()

```

```python
class a:
    def process(self):
        print("executing c")
        super().process()
class b(a):
    def process(self):
        print("executing b")
        super().process()

class c(a):
    def process(self):
        print("executing a")
        
class d(b,c):
    def process(self):
        print("executing d")
        super().process()
    
print(d.mro())
d=d()
d.process()
```

```python
class payment:
    def process(self):
        print("processing payment")
class creditcardpayment(payment):
    def process(self):
        print("processing credit card payment")
        super().process()
        print("completing credit card payment")
payment = creditcardpayment()
payment.process()
print(creditcardpayment.mro())
print(type(payment).mro())
```
