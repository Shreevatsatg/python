```python
class GameCharacter:
    game_name = "Adventure Quest"

    def __init__(self, name, level):
        self.name = name
        self.level = level
        self.health = 100

    def display_info(self):
        return f"{self.name}: Level {self.level}, Health {self.health}"

    def set_health(self, new_health):
        self.health = new_health

    def heal(self, points):
        self.health += points

    def attack(self):
        print(f"{self.name} has attacked.")

class Warrior(GameCharacter):
    def __init__(self, name, level, weapon):
        super().__init__(name, level)
        self.weapon = weapon

    def sword_attack(self):
        print(f"{self.name} attacks using {self.weapon}.")
```

```python
# Parent class invocation
character = GameCharacter("Alex", 5)
print(character.display_info())
character.attack()

# Child class invocation
warrior = Warrior("Riya", 10, "long sword")
print(warrior.display_info())  # Inherited from GameCharacter
warrior.attack()              # Inherited from GameCharacter
warrior.sword_attack()        # Defined in Warrior
```

```python
class foodorder:
    def __init__(self, cutomer_name):
        self.cutomer_name = cutomer_name
    def confirm_order(self):
        print("order confirmed for",self.cutomer_name)

person=foodorder("shree")
foodorder.confirm_order(person)
```

```python
class pizzaoder(foodorder):
    def __init__(self, cutomer_name, size, topping):
        super().__init__(cutomer_name)
        self.size=size
        self.topping=topping

    def prepare_pizza(self):
        print(f"preparing a {self.size} pizza with {self.topping}")
        
pizza=pizzaoder("shree","large","paneer")
pizza.confirm_order()
pizzaoder.prepare_pizza(pizza)
```

```python
class car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def describe_car(self):
        print(f"{self.year} {self.make} {self.model}")

class electroniccar(car):
    def __init__(self, make, model, year):
        super().__init__(make, model, year)
        self.battery_size = 70
    def describe_battery(self):
        print(f"This car has a {self.battery_size}-kWh battery.")

my_car = electroniccar("Tesla", "Model 3", 2024)
my_car.describe_car()
my_car.describe_battery()
```
