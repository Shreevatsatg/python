```python
from collections import defaultdict

int_dict=defaultdict(int)
int_dict['a']=100
int_dict['b']+=100
print(int_dict)
```

```python
int_dict.popitem();
print(int_dict)
```

```python
from collections import UserDict

class mydict(UserDict):
    def get_keys(self):
        return[keys.upper() for key in self.keys()]

#creating an object
customdict=mydict({'a':1,'b':2,'c':3})

print(customdict)
```

```python
from collections import ChainMap

dict1={'a':1, 'b':2}
dict2={'c':3, 'd':4}
dict3={'e':5, 'f':6}

chain=ChainMap(dict1,dict2,dict3)
print(chain)
print(chain['a'])
print(chain['e'])
print(chain.get('c'))
```

```python
from collections import deque

my_deque =deque([1,2,3])
print(my_deque)
my_deque.appendleft(0)
print(my_deque)
```

```python
from collections import namedtuple

person = namedtuple('Person', ['name', 'age', 'city'])
person1=person('Alice', 30, 'New York')
print(person1)
```

```python
from collections import UserList

class customlist(UserList):
    def append_twice(self,value):
        self.data.append(value)
        self.data.append(value)

    def pop_twice(self,pos):
        self.data.pop(pos)
        self.data.pop(pos)

custom_lists=customlist([1, 2, 3, 4])
print(custom_lists)
custom_lists.append_twice(5)
print(custom_lists)
custom_lists.pop_twice(0)
print(custom_lists)

```
