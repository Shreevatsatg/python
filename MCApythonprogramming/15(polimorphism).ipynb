<h2>Accessing private using name mangling<h2>

```python
class employee:
    def __init__(self,name,salary ):
        self.name = name
        self.__salary = salary

emp=employee("Ahmed", 30000)
#print("salary",emp.__salary) #not possible because its private
print("salary",emp.name)
print("salary", emp._employee__salary) #accessed using name mangling
```

```python
class stud:
    def __init__(self,name,roll_no, age):
        self.name = name
        self.__age = age
        self.__roll_no = roll_no

    def info(self):
        print("Name: ", self.name, "Age : ", self.__age,"roll no",self.__roll_no)
    
    def get_roll_no(self): #accessor
        return self.__roll_no

stu= stud('dhruv',14,21)
stu.info()
print(
stu.get_roll_no())
```
