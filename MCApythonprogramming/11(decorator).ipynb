```python
def make_pretty(func):
    def inner():
        print("I got decorated")
        func()
    return inner

@make_pretty
def ordinary():
    print("I am ordinary")
print(ordinary())
```

```python
def divide(x,y):
    print(x/y)
def outer_div(func):
    def inner(x,y):
        if y==0:
            print("Cannot divide by zero")
            return
        return func(x,y)
    return inner
divide1=outer_div(divide)
divide1(10, 2)
```

```python
def split_string(function):
    def wrapper():
        func = function()
        splitted_string = func.split()
        return splitted_string
    return wrapper

def uppercase_decorator(function):
    def wrapper():
        func = function()
        make_uppercase = func.upper()
        return make_uppercase
    return wrapper
    
@split_string
@uppercase_decorator
def say_hi():
    return 'hello there'
```
