```python
result = {x * y for x in range(4)
          if x % 2 == 0
          for y in range(5) if y > 2}

print(result)
```

```python
student=[{'name':'alice','acores':{'math':80,'science':90}},
         {'name':'bob','acores':{'math':70,'science':60}},
         {'name':'charlie','acores':{'math':90,'science':95}}
         ]
squared_scores={
    student['name']: {subject: score**2 for subject, score in student['acores'].items()}
    for student in student
}
print(squared_scores)
```
