#greatest common devisor

```python
def gcd(a,b):
    while b:
        a, b = b, a % b
    return a
print(gcd(18,19))
```

#factorial

```python
def fact(a):
    if a == 0:
        return 1
    else:
        return a * fact(a - 1)
print(fact(8))
```

#nested function

```python
def big(a,b,c):
    def fact(a):
        if a == 0:
            return 1
        else:
            return a * fact(a - 1)

    def gcd(b,c):
        while c:
            b, c = c, b % c
        return b
    return(gcd(b,c),fact(a))
print(big(8,12,18))
```
