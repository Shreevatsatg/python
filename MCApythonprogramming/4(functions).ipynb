#variable length arguments example

```python
def multiplier(*num):
    prod =1
    for i in num:
        prod=prod*i
    print("product: ",prod)
multiplier(3,5)
```

```python
def func(name,clas,*subject,**kwargs):
    print("name:",name)
    print("class:", clas)
    print("\n subjects enrolled:")
    for i in subject:
        print(i)

    print("\n other details:" )
    for key, value in kwargs.items():
        print(key, ":", value)
func("John", 10, "Math", "Science", age=15, city="New York")
```

```python
def functio(*args):
    total=0
    for val in args:
        total=total+val
    return total
ans=functio(1,2,3,4)
print(ans);
```

```python
def functio(*args):
    total=args[0]+args[1]+args[2]
    return total
ans=functio(1,2,3,4)
print(ans);
```
